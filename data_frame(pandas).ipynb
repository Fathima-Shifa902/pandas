```python
#create a dataframe
import pandas as pd
import numpy as np
datas={
    "name":["alice","bob","charlie","david","emma"],
    "age":[20,21,19,22,20],
    "marks":[85,72,np.nan,90,65],
    "attendance":[95,80,75,np.nan,88],
    "city":["kochi","delhi","mumbai","kochi","chennai"]
}
df=pd.DataFrame(datas)
print(df)
```

```python
#first rows print cheyann-head()
#last rows print cheyann-tail()
print("\n first 3 rows:")
print(df.head(3))

print("\n last 2 rows:")
print(df.tail(2))
```

```python
#shape-use for datas arrangement
print(df.shape)
print("\t.................\t")
print(df.columns)
print("\t.................\t")
print(df.dtypes)
#accessing columns
print("\t.................\t")
print(df[["name","marks"]])
print("\t.................\t")
print(df["age"])
```

```python
#slicing
print("\n first 3 rows:")
print(df.iloc[0:3])             #integer based,ith cmplte kittulla ,last elements verulla
print("\nrows 1 to 3,name and marks:")
print(df.loc[1:3,["name","marks"]])     #label or position based,complete kittum,last elements verum.
print("\t.................\t")
```

```python
#filtering
print("\n students with marks greater than 75:")
print(df[df["marks"]>75])
print("\t.................\t")
print("\n students from kochi:")
print(df[df["city"]=="kochi"])
print("\t.................\t")
print("\n students with marks>70 AND attendence>80:")
print(df[(df["marks"]>70) & (df["attendance"]>80)])
print("\t.................\t")
```

```python
#mathematical operations
print("\nmarks + 5:")
print(df["marks"]+5)
print("\nmarks * 2:")
print(df["marks"]*2)
print("\n average marks:")
print(df["marks"].mean())
print("\n minimum marks:")
print(df["marks"].min())
print("\n maximum marks:")
print(df["marks"].max())
print("\n total marks:")
print(df["marks"].sum())
```

```python
#sorting
print("\n sorted by marks:")
print(df.sort_values("marks"))

print("\n sorted by marks - highest first:")
print(df.sort_values("marks",ascending=False))
```

```python
#renaming columns
df_renamed=df.rename(columns={"name":"student_name","attendance":"attendance_percentage"})
print(df_renamed)
```

```python
#checking missing values
print("\n missing values :")
print(df.isna())
print("\n total missing values:")
print(df.isnull().sum())
print(df.dropna())
```

```python
#forward fill -null ulla value remove akki athinte mele ulla no  kond fill akkum.
df_forward=df.copy()
df_forward=df_forward.ffill()
print("\n forward fill:")
print(df_forward)
```

```python
#backward fill -null ulla value remove akki athinte thazhe ulla no kond fill akkum.
df_backward=df.copy()
df_backward=df_backward.bfill()
print("\n backward fill:")
print(df_backward)
```
