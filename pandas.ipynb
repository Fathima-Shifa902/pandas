**loc** **and** **iloc**

iloc=position type

loc=manual index based

```python
data=[1,2,3,4]
s=pd.Series(data,index=['a','b','c','d'])
print(s.loc['b'])
```

```python
data=[1,2,3,4]
s=pd.Series(data)
print(s.iloc[3])
```

```python
#slicing
data=[10,20,30,40]
s=pd.Series(data,index=['a','b','c','d'])
print(s.loc['a':'b'])
print(s.iloc[:2])
```

```python
import pandas as pd
sales = pd.Series(
    [1200,850,1500,950,2000],
    index=["monday","tuesday","wednesday","thursday","friday"]
)
print(sales)
print(sales.max())
print(sales.sum())
print(sales.min())
print(sales[sales<1000])
print(sales[sales>1000])
```
